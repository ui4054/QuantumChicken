# 📝 Punto 1: Generar estados cuánticos aleatorios

**Pregunta:**  
Implementa una función que genere **n estados cuánticos aleatorios** en un espacio de dimensión `dim = 2`.  
Cada estado debe estar **normalizado** y representado como un vector complejo.

**Entrada:**
- `n`: número de estados a generar (ej. `n= numero de estados`)  
- `dim`: dimensión del sistema cuántico (ej. `dim=2`)  
- `seed`: semilla opcional para reproducibilidad  

**Salida esperada:**  
Un **diccionario** con la siguiente estructura:
- **Clave:** índice del estado (`0 ... n-1`)  
- **Valor:** vector de estado (numpy array complejo, normalizado)  

**Ejemplo de modelo de salida:**  
```python
{
    0: array([0.707+0.j   , 0.0+0.707j]),
    1: array([0.866+0.j   , 0.0+0.5j]),
    2: array([0.612+0.354j, 0.707+.j]),
    ...
}


````python
# ---------------------------
# Punto 1: Generar 100 estados cuánticos aleatorios
# ---------------------------
def punto1_generar_estados(n=100, dim=2, seed=None):
    if seed is not None:
        np.random.seed(seed)
    #tu código aqui
    estados = {i: RandomState(dim=dim, normalized=True, real=False, seed=None) for i in range(n)}
    # fin de tu código
    return estados
````

# 📝 Punto 2: Calcular promedios y encontrar el estado representativo

**Pregunta:**  
A partir de un conjunto de estados cuánticos generados en el **Punto 1**, implementa una función que:  
1. Calcule las probabilidades de medir en $|0\rangle$ y $|1\rangle$ para cada estado usando `Sampler`.  
2. Obtenga el **promedio global** de estas probabilidades.  
3. Encuentre el **estado representativo**, es decir, aquel cuya distribución esté más cercana al promedio global.  

---

**Entrada:**
- `estados`: diccionario de estados (salida del Punto 1)  
- `n_shots`: número de repeticiones de muestreo por estado (ej. `10`)  
- `seed`: semilla opcional para reproducibilidad  

---

**Salida esperada:**  
Un **diccionario** con la siguiente estructura:
- `"avg_probs"`: vector numpy con el promedio global de probabilidades en $|0\rangle$ y $|1\rangle$  
- `"best_index"`: índice del estado más cercano al promedio  
- `"best_state"`: vector de estado correspondiente al índice representativo  

**Ejemplo de modelo de salida:**  
```python
{
    "avg_probs": array([0.55, 0.45]),
    "best_index": 3,
    "best_state": array([0.707+0.j, 0.   +0.707j])
}


````python
# ---------------------------
# Punto 2: Calcular promedios y encontrar el estado representativo
# ---------------------------
def punto2_promedios_y_mejor(estados, n_shots=100, seed=None):
    probs = []
    for i, vec in estados.items():
        # tu código aqui
        res = Sampler(vec, n_shots=n_shots, seed=seed)
        p0, p1 = res["probs"][0], res["probs"][1]
        probs.append((i, [p0, p1]))
        # hasta aquí tu código

    # Calcular promedio global
    avg_p0 = np.mean([p[1][0] for p in probs]) # tu codigo aqui
    avg_p1 = np.mean([p[1][1] for p in probs]) #tu codigo aqui
    avg = np.array([avg_p0, avg_p1])

    # Inicializa el estado más cercano al promedio
    min_dist = float("inf")
    best_index = None
    best_state = None

    # Identificar el estado más cercano al promedio
    for idx, p in probs:
        dist = np.linalg.norm(avg - np.array(p))
        if dist < min_dist:
            min_dist = dist
            best_index = idx
            best_state = estados[idx]


    return {
        "avg_probs": avg,
        "best_index": best_index,
        "best_state": best_state
    }
````

# 📝 Punto 3: Refinamiento con Estimator

**Pregunta:**  
Usando el **estado representativo** encontrado en el **Punto 2**, aplica un proceso de refinamiento con `Estimator` que:  
1. Ejecute múltiples simulaciones con un número creciente de *shots*.  
2. Verifique si el **error estándar** (`stderr_sampled`) cae por debajo de la tolerancia (`tol`).  
3. Devuelva la primera configuración de *shots* que cumpla la condición porcentaje de desviación estándar de **`0.05`**

---

**Entrada:**
- `best_state`: vector de estado representativo obtenido en el Punto 2.  
- `max_shots`: número máximo de *shots* permitidos (ej. `500000`).  
- `tol`: tolerancia de error estándar requerida (ej. `0.002`).  
- `seed`: semilla opcional para reproducibilidad.  

---

**Salida esperada:**  
Un **diccionario** con los siguientes campos:  
- `"n_shots_used"`: cantidad de *shots* utilizados cuando se alcanzó la tolerancia.  
- `"probs_theorical"`: probabilidades teóricas exactas del estado.  
- `"freqs_sampled"`: frecuencias de muestreo observadas.  
- `"stderr_sampled"`: error estándar de la estimación.  

Si no se alcanza la tolerancia dentro de `max_shots`, la función devuelve `None`.  

---

**Ejemplo de modelo de salida:**  
```python
{
    "n_shots_used": 250,
    "probs_theorical": array([0.50, 0.50]),
    "freqs_sampled": array([0.492, 0.508]),
    "stderr_sampled": array([0.0159, 0.0159])
}


````python
# ---------------------------
# Punto 3: Refinamiento con Estimator
# ---------------------------
def punto3_refinamiento(best_state, max_shots=500000, tol=0.002, seed=None):
    n_shots = 10
    while n_shots <= max_shots:
        # tu codigo aqui
        result = Estimator(best_state, n_shots=n_shots, seed=seed) # tu codigo aqui
        # hasta aqui tu código

        if np.all(result["stderr_sampled"] <= tol):

            return { # tu código aqui
                "n_shots_used": n_shots,
                "probs_theorical": result["probs_theorical"],
                "freqs_sampled": result["freqs_sampled"],
                "stderr_sampled": result["stderr_sampled"]
                } # hasta aqui tu código

        n_shots += 10  # doblar los shots cada iteración

    return None  # si no se alcanza la tolerancia
````

**Verifica tu Respuesta de Punto 3:**  
Ejecuta el grader para ir sumando tu promedio

````python
metrics = punto3_refinamiento(probs_idx_state["best_state"], seed=None)
graderReal.check_punto3(metrics)
print("Porcentaje de Desviación Estandar en |0> y |1> ",metrics['stderr_sampled'])
print("Nota Acumulada puntos 1 , 2 y 3 ",graderReal.grade())
````
